##  1. Data Cleaning

```python
import pandas as pd
df = pd.read_csv('synthetic_dmri_dataset.csv')
print("Initial shape:", df.shape)
df.head(3)
```

**Remove Duplicate Data**  
  Use `.drop_duplicates()` to avoid redundant entries.

```python
print("\nBefore removing duplicates:", df.shape)
df = df.drop_duplicates()
print("After removing duplicates:", df.shape)
```

**Handle Missing Values**  
  Fill, interpolate, or remove based on context.


```python
# Show missing values per column
print("\nMissing values before handling:\n", df.isnull().sum())

# Strategy: Fill numeric columns with median, categorical with mode
for col in df.columns:
    if df[col].dtype == 'object':
        df[col].fillna(df[col].mode()[0], inplace=True)
    else:
        df[col].fillna(df[col].median(), inplace=True)

print("Missing values after handling:\n", df.isnull().sum())
```

**Correct Data Types**  
  E.g., convert dates, string numbers to integers.


```python
print("\nData types before:")
print(df.dtypes)

# Example: Ensure 'age' is integer
df['age'] = df['age'].astype(int)

print("\nData types after:")
print(df.dtypes)
```

**Remove Unwanted Rows**  
  Filter out artifacts or irrelevant samples.


```python
print("\nShape before removing invalid rows:", df.shape)
df = df[df['brain_volume'] > 500]  # Example rule
print("Shape after:", df.shape)
```

**Handle Inconsistent Categorical Data**  
  Standardize (e.g., "Male" vs. "M").


```python
print("\nUnique gender values before:", df['gender'].unique())

# Normalize gender values
df['gender'] = df['gender'].replace({'M': 'Male', 'F': 'Female', 'male': 'Male', 'female': 'Female'})

print("Unique gender values after:", df['gender'].unique())
```

**Validate Data Integrity**  
  E.g., correct mismatched IDs or modality labels.

```python
print("\nAge range before integrity check:", df['age'].min(), "-", df['age'].max())
df = df[(df['age'] >= 40) & (df['age'] <= 100)]
print("Age range after:", df['age'].min(), "-", df['age'].max())
```

**Remove Irrelevant Data**  
Drop unused columns or metadata.

```python
print("\nColumns before dropping irrelevant ones:\n", df.columns.tolist())

# Drop unnecessary columns
irrelevant = ['irrelevant_column'] if 'irrelevant_column' in df.columns else []
df.drop(columns=irrelevant, inplace=True)

print("Columns after dropping:\n", df.columns.tolist())
```

**Handle Mixed Data Types in a Column**  
  Ensure type uniformity for features.

```python
# Introduce a fault for demonstration
df.loc[0, 'age'] = '55 years'
print("\nExample faulty entry:", df.loc[0, 'age'])

# Fix it
df['age'] = df['age'].astype(str).str.extract('(\d+)').astype(float).astype(int)

print("Corrected entry:", df.loc[0, 'age'])
```

## 2. Data Validation & Consistency


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import zscore
from sklearn.ensemble import IsolationForest

# Reload clean dataset
df = pd.read_csv("synthetic_dmri_dataset_cleaned.csv")

# Helper
def print_range(col):
    print(f"{col} range: {df[col].min()} to {df[col].max()}")
```

- [ ] **Check for Data Consistency**  
  E.g., no negative ages or brain volumes.


```python
# E.g., No negative age or volumes, features within real-world ranges

print("\n🔍 Consistency checks:")
assert (df["age"] >= 0).all(), "Negative age found!"
assert (df["brain_volume"] > 500).all(), "Suspiciously low brain volume!"

dmri_cols = [col for col in df.columns if col not in ["subject_id", "gender", "diagnosis", "age"]]

for col in dmri_cols:
    if (df[col] < 0).any():
        print(f"Warning: Negative values in {col}")
```

- [ ] **Check for Data Skewness**  
  Visualize feature distributions.


```python
print("\n📊 Skewness:")
skewed = df[dmri_cols].skew().sort_values(ascending=False)
print(skewed)

# Plot top 6 skewed features
top_skewed = skewed.head(6).index.tolist()

df[top_skewed].hist(bins=20, figsize=(12, 6), layout=(2, 3))
plt.suptitle("Top Skewed dMRI Feature Distributions")
plt.tight_layout()
plt.show()
```

- [ ] **Detect and Handle Data Anomalies**  
  Use z-scores, isolation forests, etc.


```python
# Using Z-score
print("\n⚠️ Anomaly detection (Z-score method):")
z_scores = np.abs(zscore(df[dmri_cols]))
outlier_rows_z = (z_scores > 3).any(axis=1)
print(f"Z-score: {outlier_rows_z.sum()} subjects flagged as potential outliers")

# Isolation Forest
print("\n⚠️ Anomaly detection (Isolation Forest):")
iso = IsolationForest(contamination=0.01, random_state=42)
outliers_iso = iso.fit_predict(df[dmri_cols])
df["is_outlier"] = outliers_iso
print("Isolation Forest flagged:", (df["is_outlier"] == -1).sum(), "subjects")

# Optional: remove them
# df = df[df["is_outlier"] != -1]
```

- [ ] **Detect and Handle Data Leakage**  
  Make sure no test info leaks into training.


```python
print("\n🧪 Data Leakage Checks:")
# Check for duplicates between test/train? Not splitting yet, but check label/feature correlation
correlation = df[dmri_cols + ["diagnosis"]].apply(lambda x: pd.factorize(x)[0]).corr()
high_corrs = correlation["diagnosis"].abs().sort_values(ascending=False)
print("Features highly correlated with diagnosis (possible leakage risk):")
print(high_corrs[1:6])  # skip self-correlation
```

- [ ] **Validate Label Accuracy**  
  Especially critical in clinical datasets.

```python
print("\n🧪 Label-Diagnosis Checks:")
# Check brain volume by diagnosis (as an example clinical relationship)
sns.boxplot(x="diagnosis", y="brain_volume", data=df)
plt.title("Brain Volume by Diagnosis")
plt.show()

# Check age by diagnosis
sns.boxplot(x="diagnosis", y="age", data=df)
plt.title("Age Distribution by Diagnosis")
plt.show()
```

## 3. Data Transformation


```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.neighbors import LocalOutlierFactor
from imblearn.over_sampling import SMOTE

df = pd.read_csv('synthetic_dmri_dataset.csv')
print("Initial shape:", df.shape)
df.head(3)
```

- [ ] **Standardize Data**  
  E.g., unit conversion, consistent naming.

```python
# Example: Rename columns to lowercase and replace spaces with underscores
df.columns = [col.strip().lower().replace(" ", "_") for col in df.columns]

# Example: Convert b-values from s/mm^2 to consistent scale (if applicable)
if 'b_value' in df.columns:
    df['b_value'] = df['b_value'].apply(lambda x: x * 1000 if x < 10 else x)  # Convert from s/sqmm to s/mm^2

print("After standardizing columns:")
print(df.columns.tolist())
```

- [ ] **Normalize Data**  
  Scale values (e.g., min-max, z-score).

```python
# Identify numeric columns to normalize (excluding target/categorical)
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
numeric_cols = [col for col in numeric_cols if col != 'target']  # Exclude target if exists

scaler = MinMaxScaler()
df[numeric_cols] = scaler.fit_transform(df[numeric_cols])

print("\nNormalized numerical columns:")
print(df[numeric_cols].head(3))
```

- [ ] **Encode Categorical Variables**  
  One-hot, label encoding, etc.

```python
# Label encoding for binary vars, one-hot for multi-class
cat_cols = df.select_dtypes(include=['object', 'category']).columns.tolist()

# Apply Label Encoding or One-Hot
for col in cat_cols:
    if df[col].nunique() == 2:
        le = LabelEncoder()
        df[col] = le.fit_transform(df[col])
    else:
        df = pd.get_dummies(df, columns=[col], drop_first=True)

print("\nAfter encoding categorical variables:")
print("Current shape:", df.shape)
```

- [ ] **Filter Outliers**  
  Use domain rules or statistical thresholds.

```python
# Use Local Outlier Factor (LOF) as an example
lof = LocalOutlierFactor(n_neighbors=20, contamination=0.05)
outliers = lof.fit_predict(df[numeric_cols])
df = df[outliers != -1]

print("\nAfter filtering outliers:")
print("Shape after removing outliers:", df.shape)
```

- [ ] **Handle Sparse Data**  
  E.g., in connectome or gene-based features.

```python
# Drop columns that are >95% zeros
sparse_cols = [col for col in numeric_cols if (df[col] == 0).mean() > 0.95]
df.drop(columns=sparse_cols, inplace=True)

print("\nAfter removing sparse features:")
print("Removed columns:", sparse_cols)
print("Shape:", df.shape)
```

- [ ] **Handle Noisy Data**  
  Apply smoothing or noise reduction.

```python
# For illustration, apply rolling mean to a diffusion feature if time series-like (fake example)
if 'fractional_anisotropy' in df.columns:
    df['fa_smooth'] = df['fractional_anisotropy'].rolling(window=3, min_periods=1).mean()

print("\nAfter smoothing noisy data (if time-based):")
df[['fractional_anisotropy', 'fa_smooth']].head(5)
```

- [ ] **Handle Imbalanced Data**  
  Use oversampling or SMOTE if needed.

```python
if 'target' in df.columns:
    X = df.drop('target', axis=1)
    y = df['target']

    smote = SMOTE(random_state=42)
    X_res, y_res = smote.fit_resample(X, y)

    df = pd.concat([pd.DataFrame(X_res, columns=X.columns), pd.Series(y_res, name='target')], axis=1)

    print("\nAfter handling imbalanced data with SMOTE:")
    print(df['target'].value_counts())
```

- [ ] **Aggregate Data**  
  E.g., region-based summary from voxel data.


```python
# Simulate aggregation by averaging voxel values per region
region_prefixes = set([col.split('_')[0] for col in df.columns if 'voxel' in col])
for region in region_prefixes:
    region_voxels = [col for col in df.columns if col.startswith(region+'_voxel')]
    df[region + '_mean'] = df[region_voxels].mean(axis=1)
    df.drop(columns=region_voxels, inplace=True)

print("\nAfter aggregating voxel data by region:")
print("Shape:", df.shape)
```


## 4. Feature Engineering & Selection


- [ ] **Detect and Remove Multicollinearity**  
  Use correlation matrix or VIF.

```python
import seaborn as sns
import matplotlib.pyplot as plt
from statsmodels.stats.outliers_influence import variance_inflation_factor

# Step 1: Correlation matrix heatmap
plt.figure(figsize=(10, 8))
corr = df.corr()
sns.heatmap(corr, cmap="coolwarm", annot=False, fmt=".2f", xticklabels=False, yticklabels=False)
plt.title("Feature Correlation Matrix")
plt.show()

# Step 2: Drop features with high correlation (e.g., > 0.95)
threshold = 0.95
upper = corr.where(np.triu(np.ones(corr.shape), k=1).astype(bool))
drop_cols_corr = [column for column in upper.columns if any(upper[column].abs() > threshold)]
df.drop(columns=drop_cols_corr, inplace=True)

print("\nDropped highly correlated features (correlation > 0.95):")
print(drop_cols_corr)

# Step 3: VIF (optional but more precise)
X = df.select_dtypes(include=[np.number]).drop(columns=['target'], errors='ignore')  # Remove target

vif_data = pd.DataFrame()
vif_data['feature'] = X.columns
vif_data['VIF'] = [variance_inflation_factor(X.values, i) for i in range(len(X.columns))]
print("\nVariance Inflation Factor (VIF):")
print(vif_data[vif_data['VIF'] > 10])  # Threshold commonly set to 5 or 10
```

- [ ] **Apply Feature Extraction**  
  PCA, ICA, CNN bottlenecks, etc.


```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# Standardize again for PCA
features = df.select_dtypes(include=[np.number]).drop(columns=['target'], errors='ignore')
scaler = StandardScaler()
X_scaled = scaler.fit_transform(features)

# Apply PCA
pca = PCA(n_components=0.95)  # retain 95% of variance
X_pca = pca.fit_transform(X_scaled)

print("\nPCA: Number of components to explain 95% variance:", pca.n_components_)

# Replace df with PCA features + target
df_pca = pd.DataFrame(X_pca, columns=[f"PC{i+1}" for i in range(pca.n_components_)])
if 'target' in df.columns:
    df_pca['target'] = df['target'].values

print("Shape after PCA:", df_pca.shape)
df_pca.head(3)
```

- [ ] **Generate New Features**  
  E.g., brain volume ratios, asymmetries.

```python
# Example: Left-right asymmetry in FA or MD
if all(col in df.columns for col in ['fa_left', 'fa_right']):
    df['fa_asymmetry'] = (df['fa_left'] - df['fa_right']) / (df['fa_left'] + df['fa_right'] + 1e-6)

if all(col in df.columns for col in ['volume_left', 'volume_right']):
    df['volume_ratio'] = df['volume_left'] / (df['volume_right'] + 1e-6)

print("\nAfter generating domain-informed features (asymmetry and ratios):")
print(df[['fa_asymmetry', 'volume_ratio']].head(3))
```
