## 2. The Logistic Regression Model
- **Logistic Function (Sigmoid)**:
    - Definition: 
        $ \sigma(z) = \frac{1}{1 + e^{-z}} $
    - Properties: Maps any real-valued number to a value between 0 and 1.
    - Why we use the sigmoid: It transforms the linear regression output into a probability that can be interpreted in classification tasks.

- **Decision Boundary**:
    - The decision boundary is where the model switches classification (e.g., 0.5 threshold).
    - **Thresholding**: How we set a threshold (e.g., 0.5) to classify the predictions.

## 3. Logistic Regression Equation
- **Model Representation**:
    $ P(y=1 | X) = \sigma(X \cdot \beta) $


    Where $ ( \beta )$ is the vector of coefficients.
    - **Interpretation** of the coefficients and how they relate to odds ratios in neuroimaging context (e.g., likelihood of having Alzheimer's based on biomarkers).

## 4. Training Logistic Regression
- **Cost Function**:
    - Log-Loss / Binary Cross-Entropy:
    
    $\text{Cost} = - \frac{1}{N} \sum_{i=1}^{N} \left[ y_i \log(h(x_i)) + (1 - y_i) \log(1 - h(x_i)) \right] $
    
    
    - Why minimizing log-loss is crucial for training.
    - Gradient descent optimization: How the model’s coefficients are updated.

### 3.4 Hyperparameters?!

- **`C`** (Regularization strength) – Controls the amount of regularization.
- **`solver`** – Determines the algorithm to solve the problem (choose based on dataset size and type).
- **`penalty`** – Regularization method (L1, L2, ElasticNet, or None).
- **`max_iter`** – Iteration limit, controls when optimization stops.
- **`multi_class`** – For multi-class problems, typically 'ovr' (One-vs-Rest) is used.

#### 1. **`C` (Regularization Strength)**

- **Definition**: The `C` parameter controls the strength of regularization applied to the model. It determines the trade-off between fitting the model to the data and keeping the model's coefficients small to prevent overfitting.
- **How it works**:
    - A small value of `C` means **strong regularization**, where the model will favor simpler solutions (coefficients close to zero) and may not fit the data as well. This helps to prevent overfitting when the data is noisy.
    - A large value of `C` means **weaker regularization**, where the model will allow larger coefficients and may overfit the data, especially if there is high variance or noise.
- **Neuroimaging Context**:
    - In neuroimaging, where features like **tract volume, FA mean**, or **WMH load** can have varying importance and noise levels, adjusting `C` helps balance the model's ability to generalize across subjects while keeping the features' relationships meaningful. Strong regularization can be helpful in case of highly correlated features (e.g., FA and MD).

---

#### 2. **`solver` (Optimization Algorithm)**

- **Definition**: The `solver` parameter specifies which optimization algorithm to use for training the model.
- **Common solvers**:
    - **`liblinear`**: Good for small datasets and binary classification problems. It supports both L1 and L2 regularization.
    - **`lbfgs`**: Suitable for larger datasets and supports only L2 regularization. This is often used when you have many features.
    - **`newton-cg`**: A second-order method that works well for large datasets and multi-class classification.
    - **`saga`**: A variant of `lbfgs` suitable for large datasets and supports both L1 and L2 regularization.
- **Neuroimaging Context**:
    - For neuroimaging datasets with a large number of features (e.g., brain volumes, FA, MD, etc.), solvers like `lbfgs` or `saga` are preferable as they can handle large-scale, high-dimensional datasets efficiently. If the task is binary classification (e.g., CN vs. AD), `liblinear` can be a good option for small datasets.

---

#### 3. **`penalty` (Regularization Type)**

- **Definition**: The `penalty` parameter determines the type of regularization applied to the model. It helps to control model complexity by penalizing larger weights (coefficients).
- **Common penalties**:
    - **`l2`** (default): L2 regularization, which adds a penalty proportional to the square of the coefficients. It prevents overfitting by shrinking large coefficients without setting them to zero.
    - **`l1`**: L1 regularization, which can produce sparse solutions by forcing some coefficients to be exactly zero. This can help with feature selection in high-dimensional neuroimaging data.
    - **`elasticnet`**: Combines both L1 and L2 regularization. This can be useful when you want to strike a balance between sparsity and model complexity.
    - **`none`**: No regularization.
- **Neuroimaging Context**:
    - In neuroimaging, where the number of features is often much larger than the number of samples, **L1 regularization** (Lasso) can help identify the most relevant features (e.g., specific brain regions or tracts) by forcing irrelevant features (e.g., less informative metrics like `NDI` or `ODI`) to have zero coefficients.
    - **L2 regularization** (Ridge) is usually a safe default, especially when most features are expected to contribute to the model.

---

#### 4. **`max_iter` (Maximum Number of Iterations)**

- **Definition**: The `max_iter` parameter controls the maximum number of iterations the solver will run to converge.
- **How it works**:
    - If the solver does not converge to a solution within the given number of iterations, it will stop and raise a warning.
    - Increasing this value ensures that the model has enough iterations to converge, especially when dealing with complex or large datasets.
- **Neuroimaging Context**:
    - Neuroimaging data often contains highly correlated features and complex relationships between them, which can slow down the convergence process. You may need to increase `max_iter` to ensure convergence, especially when using solvers like `saga` or `newton-cg`.

---

#### 5. **`multi_class` (Handling Multi-Class Classification)**

- **Definition**: The `multi_class` parameter determines how the logistic regression model handles multi-class classification problems (e.g., CN, MCI, AD).
- **Common values**:
    - **`ovr`** (One-vs-Rest): Treats the multi-class problem as multiple binary classification problems (one classifier per class).
    - **`multinomial`**: Optimizes the multi-class loss directly. This is often more efficient for problems where multiple classes are involved and the classes are not easily separable.
- **Neuroimaging Context**:
    - In neuroimaging, where the task is often to classify multiple stages of Alzheimer's (e.g., CN, MCI, AD), the **One-vs-Rest** approach (`ovr`) is usually fine unless the dataset is highly imbalanced. If you expect more complex relationships between the classes, using **`multinomial`** could be beneficial.

---

#### 6. **`class_weight` (Handling Class Imbalance)**

- **Definition**: The `class_weight` parameter adjusts the weight of each class during training to handle class imbalances (e.g., more CN samples than AD).
- **Common values**:
    - **`balanced`**: Automatically adjusts weights inversely proportional to class frequencies, useful for imbalanced datasets.
    - **`None`**: No adjustment (default).
- **Neuroimaging Context**:
    - In neuroimaging studies, the number of subjects in each class (e.g., CN, MCI, AD) may be imbalanced. Setting `class_weight='balanced'` ensures that the model does not bias predictions towards the majority class and improves the model's ability to predict minority classes like AD or MCI.

---

#### 7. **`fit_intercept` (Intercept Term)**

- **Definition**: Whether to include an intercept term in the model or not.
- **Values**:
    - **`True`** (default): Includes an intercept term.
    - **`False`**: No intercept term is included.
- **Neuroimaging Context**:
    - In most neuroimaging applications, an intercept term is usually necessary, as it allows the model to adjust for baseline differences across subjects (e.g., age, gender, scanner).

---

#### 8. **`tol` (Tolerance for Stopping Criterion)**

- **Definition**: The `tol` parameter controls when the solver should stop based on the change in the cost function (i.e., when further optimization won't improve the result significantly).
- **Values**:
    - A small positive float, such as `1e-4`, `1e-5`.
- **Neuroimaging Context**:
    - In neuroimaging, where the data can be noisy and have a large number of features, adjusting `tol` can help control the precision of the model’s convergence. A smaller `tol` can ensure the model converges with higher precision, but at the cost of more computation.

---

#### 9. **`warm_start` (Reusing Previous Solution)**

- **Definition**: When set to `True`, the `warm_start` parameter allows you to reuse the solution of the previous call to `fit()` and add more iterations or data without starting from scratch.
- **Values**:
    - **`True`**: Reuses the previous solution.
    - **`False`** (default): Starts fresh.
- **Neuroimaging Context**:
    - If you're training the model incrementally with new data or fine-tuning a model with additional features, setting `warm_start=True` can save computation time by reusing previous results.
