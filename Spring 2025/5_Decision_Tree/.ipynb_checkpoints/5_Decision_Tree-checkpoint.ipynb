# Decision Tree

## What is a Decision Tree?

A Decision Tree is a machine learning model that looks like a flowchart.

It makes decisions by asking questions at each step and splitting the data based on the answers.


To classify or predict something by splitting data into groups that are as pure as possible.



**Why is it called a "Tree"?** 

Because it starts at the root and branches out like this:

```python
              [Age > 60?]
              /       \
         Yes /         \ No
            /           \
  [BP > 140?]         Healthy
     /    \
  Yes      No
AD Risk    MCI
```

Each node = a question (split condition).

Each branch = the result of the question.

Each leaf = the final decision (prediction).

**How Does a Decision Tree Work?**

It builds the tree by splitting the data again and again.

At each step, it chooses the best question using a mathematical score.

| Feature            | Description                               |
| ------------------ | ----------------------------------------- |
| Model Type         | Supervised (classification or regression) |
| Structure          | Tree-like with nodes and leaves           |
| Splitting Criteria | Gini or Entropy                           |
| Overfitting Risk   | High                                      |
| Interpretability   | Excellent                                 |
| Main Use Cases     | Medical decisions, loan approval, etc.    |


**Classification**: Deciding which group something belongs to.
→ Example: Is the patient Healthy, has MCI, or AD?

**Regression**: Predicting a number.
→ Example: Predict a person’s memory score.

***How Pure Is a Group?***

We use measures of impurity:

A. Gini Impurity (Default)

        Lower is better (0 = pure)

        Gini = 0 means all in one class

B. Entropy (Information Gain)

        Based on information theory

        Slower but sometimes more accurate

```python
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report

X, y = load_iris(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
clf = DecisionTreeClassifier(criterion='gini', max_depth=3, random_state=42)
clf.fit(X_train, y_train)
y_pred = clf.predict(X_train)
print(classification_report(y_test, y_train))
```

### Types of Decision Trees



1. ***Basic Decision Tree (CART)***
   
        CART = Classification and Regression Tree.

        Can be used for classification or regression.

3. ***Random Forest***
   
        A group (or "forest") of decision trees.

        Each tree votes, and the majority wins.

5. ***Gradient Boosting Trees***
   
        Trees are built one after another.

        Each new tree fixes mistakes made by the last.

        Very powerful and popular (used in XGBoost, LightGBM).

## Decision Tree Hyperparameters to Tune

| Parameter           | Meaning                                       |
| ------------------- | --------------------------------------------- |
| `criterion`         | How to split? (`gini` or `entropy`)           |
| `max_depth`         | How deep the tree can go                      |
| `min_samples_split` | Minimum samples to split a node               |
| `min_samples_leaf`  | Minimum samples in a leaf node                |
| `max_features`      | Max number of features to consider at a split |
| `class_weight`      | Helps with imbalanced data (`'balanced'`)     |
| `random_state`      | Seed for reproducibility                      |


**max_depth**

Limits how deep the tree can grow. Deeper trees = more complex = higher risk of overfitting.

```python
clf = DecisionTreeClassifier(max_depth=2)
```

**min_samples_split**

Minimum number of samples required to split a node. Higher values prevent the tree from growing too deep.

```python
DecisionTreeClassifier(min_samples_split=10)
```

**max_leaf_nodes**

Limits the number of final decision points (leaf nodes). Helps keep the model simple.


```python
DecisionTreeClassifier(max_leaf_nodes=5)
```

**max_features**

Limits the number of features considered at each split. Common choices:

'sqrt' → good for classification

'log2'

None (default) → use all features

```python
DecisionTreeClassifier(max_features='sqrt')
```

**criterion**

Defines the metric used to decide splits:

'gini' → Gini impurity (default)

'entropy' → Information gain

```python
 DecisionTreeClassifier(criterion='entropy')
```

**class_weight**

Helps deal with imbalanced classes. Use 'balanced' to automatically adjust weights.



```python
DecisionTreeClassifier(class_weight='balanced')
```

# Random Forest

## What is a Random Forest?

A **Random Forest** is a collection (or "forest") of Decision Trees.

Each tree gives a prediction.
The forest takes a vote (for classification) or average (for regression) to make the final decision.

Random Forest is a collection of Decision Trees working together to make better predictions.

Think of it like a *team of decision trees* each one gives a vote, and the majority wins.

-------------

-------------

### How it works

1. Make many Decision Trees

            Each tree gets a random sample of the data (called bootstrapping).

            At each split, it chooses a random subset of features.

   

3. Let each tree make a prediction
   
4. Take the majority vote (for classification) or average (for regression)

This randomness makes each tree slightly different, so they don’t all make the same mistakes.

```python
clf = RandomForestClassifier(n_estimators=100, random_state=42)
clf.fit(X_train, y_train)
```

| Parameter           | Meaning                                 |             |

| `max_features`      | Max features to consider when splitting |                      |

# Other Ensemble Mdels

1. **Bagging**:
   
        Trains multiple models on different random subsets of the training data (with replacement).

        Aggregates predictions using majority voting (for classification).

        Reduces variance.

        Example: Random Forest (ensemble of decision trees).


2. **Boosting**

        Trains models sequentially, each focusing more on the mistakes of the previous one.

        Aggregates weighted predictions.

        Reduces bias and can handle complex patterns.

        Examples: AdaBoost, Gradient Boosting, XGBoost, LightGBM, CatBoost.

3. **Stacking**

        Trains several different models (e.g., SVM, kNN, decision tree), then uses another model (meta-learner) to learn how to best combine their outputs.

        More flexible but also more complex.



### Bagging = Reduces variance (e.g., Random Forest)

### Boosting = Reduces bias (e.g., AdaBoost, GBoost)

### Voting = Combines already-trained models (good when you have several strong classifiers)

### Stacking = Learns how to combine multiple models via another model

***Why Use Ensembles for Classification?***

    Reduce overfitting (especially bagging).

    Improve accuracy (especially boosting).

    Combine strengths of different models.

    More robust to noise and data variance.

| Situation                                         | Use This Ensemble                     |
| ------------------------------------------------- | ------------------------------------- |
| You want to reduce **variance** (unstable models) | **Bagging** (e.g., Random Forest)     |
| You want to reduce **bias** (underfitting)        | **Boosting** (e.g., XGBoost)          |
| You have strong, diverse models already           | **Stacking**                          |
| You have **imbalanced** data                      | Boosting (e.g., SMOTEBoost, AdaBoost) |
| Need interpretability and speed                   | Bagging (trees)                       |
